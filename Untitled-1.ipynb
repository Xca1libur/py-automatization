```python
def sum_to(n):
    number = 1
    total = 0

    while number <= n:
        total += number
        number += 1

    return total

print(sum_to(20))
```

```python
def dev_num(n):
    devisors = []
    
    for devisor in range(2, n + 1):
        while n % devisor == 0:
            n = n // devisor
            devisors.append(devisor)
            

    return devisors

print(dev_num(4050))
```

```python
def numbers_prime(n):
    divisors = []

    for number in range(1, n + 1, 1):
        if n % number == 0:
            count_div = 0
            
            for num in range(1, number + 1):
                if number % num == 0:
                    count_div += 1

            if count_div == 2:
                divisors.append(number)

    return divisors

print(numbers_prime(20)) 
```
